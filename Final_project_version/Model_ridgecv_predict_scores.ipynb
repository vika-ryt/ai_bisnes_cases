```python
import re

import joblib
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.pipeline import Pipeline
```

```python
df = pd.read_excel("/content/cases_train2 (15).xlsx")
df["Оценка"] = df["Оценка"].astype("float64")
df_test_cases = pd.read_excel("/content/Копия Test1.xlsx")
df_test_cases["Оценка"] = df_test_cases["Оценка"].astype("float64")
br = r"<br\s*/?>"
texts = df["Решение кейса"].astype(str).str.replace(
    br, " ", regex=True)
texts_test = df_test_cases["Решение кейса"].astype(str).str.replace(
    br, " ", regex=True)
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
# Модель для оценивания критерия Анализ ЦА
y_train_audience = df["ЦА"].astype(float).values
y_test_audience = df_test_cases["ЦА"].astype(float).values
model_audience = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                              sublinear_tf=True, max_features=20000)),
    ("ridge", RidgeCV(alphas=np.logspace(-2, 2, 15))),
])
pred_cv_audience = np.clip(
    cross_val_predict(model_audience, texts, y_train_audience, cv=kf), 1, 5)
print("ЦА MAE Val",
      mean_absolute_error(y_train_audience, pred_cv_audience))
model_audience.fit(texts, y_train_audience)
pred_test_audience = np.clip(model_audience.predict(texts_test), 1, 5)
print("ЦА MAE Test",
      mean_absolute_error(y_test_audience, pred_test_audience))
joblib.dump(model_audience, "lr_audience.joblib")
```

```python
# Модель для оценивания критерия Проработка решения
y_train_sol = df["Проработка решения"].astype(float).values
y_test_sol = df_test_cases["Проработка решения"].astype(float).values
model_sol = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                              sublinear_tf=True, max_features=20000)),
    ("ridge", RidgeCV(alphas=np.logspace(-2, 2, 15))),
])
pred_cv_sol = np.clip(
    cross_val_predict(model_sol, texts, y_train_sol, cv=kf), 1, 5)
print("Проработка решения MAE Val",
      mean_absolute_error(y_train_sol, pred_cv_sol))
model_sol.fit(texts, y_train_sol)
pred_test_sol = np.clip(model_sol.predict(texts_test), 1, 5)
print("Проработка решения MAE Test",
      mean_absolute_error(y_test_sol, pred_test_sol))
joblib.dump(model_sol, "lr_sol.joblib")
```

```python
# Модель для оценивания критерия Финансовая модель и метрики
y_train_finance = df["Финансовая модель и метрики"].astype(float).values
y_test_finance = (
    df_test_cases["Финансовая модель и метрики"].astype(float).values)
model_finance = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                              sublinear_tf=True, max_features=20000)),
    ("ridge", RidgeCV(alphas=np.logspace(-2, 2, 15))),
])
pred_cv_finance = np.clip(
    cross_val_predict(model_finance, texts, y_train_finance, cv=kf), 1, 5)
print("Финансовая модель и метрики MAE Val",
      mean_absolute_error(y_train_finance, pred_cv_finance))
model_finance.fit(texts, y_train_finance)
pred_test_finance = np.clip(model_finance.predict(texts_test), 1, 5)
print("Финансовая модель и метрики MAE Test",
      mean_absolute_error(y_test_finance, pred_test_finance))
joblib.dump(model_finance, "lr_finance.joblib")
```

```python
# Модель для оценивания критерия Анализ рисков
y_train_risks = df["Анализ рисков"].astype(float).values
y_test_risks = df_test_cases["Анализ рисков"].astype(float).values
model_risks = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                              sublinear_tf=True, max_features=20000)),
    ("ridge", RidgeCV(alphas=np.logspace(-2, 2, 15))),
])
pred_cv_risks = np.clip(
    cross_val_predict(model_risks, texts, y_train_risks, cv=kf), 1, 5)
print("Анализ рисков MAE Val",
      mean_absolute_error(y_train_risks, pred_cv_risks))
model_risks.fit(texts, y_train_risks)
pred_test_risks = np.clip(model_risks.predict(texts_test), 1, 5)
print("Анализ рисков MAE Test",
      mean_absolute_error(y_test_risks, pred_test_risks))
joblib.dump(model_risks, "lr_risks.joblib")
```

```python
# Модель для оценивания критерия Доказательства
y_train_proves = df["Доказательства"].astype(float).values
y_test_proves = df_test_cases["Доказательства"].astype(float).values
model_proves = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3,
                              sublinear_tf=True, max_features=20000)),
    ("ridge", RidgeCV(alphas=np.logspace(-2, 2, 15))),
])
pred_cv_proves = np.clip(
    cross_val_predict(model_proves, texts, y_train_proves, cv=kf), 1, 5)
print("Доказательства MAE Val",
      mean_absolute_error(y_train_proves, pred_cv_proves))
model_proves.fit(texts, y_train_proves)
pred_test_proves = np.clip(model_proves.predict(texts_test), 1, 5)
print("Доказательства MAE Test",
      mean_absolute_error(y_test_proves, pred_test_proves))
joblib.dump(model_proves, "lr_proves.joblib")
```

```python
# Общая оценка на тесте
pred_test_total = (pred_test_audience + pred_test_sol + pred_test_finance
                   + pred_test_risks + pred_test_proves) / 5
y_test_total = df_test_cases["Оценка"].astype(float).values
print("Оценка MAE Test", mean_absolute_error(y_test_total, pred_test_total))
```

```python
# Общая оценка на валидации
pred_cv_total = (pred_cv_audience + pred_cv_sol + pred_cv_finance
                 + pred_cv_risks + pred_cv_proves) / 5
y_train_total = df["Оценка"].astype(float).values
print("Оценка MAE Val", mean_absolute_error(y_train_total, pred_cv_total))
```
